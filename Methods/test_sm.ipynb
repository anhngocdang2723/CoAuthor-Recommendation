```python
import numpy as np
def check_nan(matrix):
    nan_indices = np.argwhere(np.isnan(matrix))
    
    if nan_indices.size == 0:
        return True, None
    else:
        return False, nan_indices
```

```python
def check_U_supervised(matrix):
    for row in matrix:
        if row[0] <= row[1]:
            return False 
    return True
```

```python
def check_pos_label(matrix,y):
    for row in range(len(matrix)):
        if y[row]==1:
            return True
    return False
```

```python
def is_subarray(sub_array, parent_array):
    is_sub_array = all(element in parent_array for element in sub_array)
    if(is_sub_array):
        return True

    # Nếu không tìm thấy
    return False

# Ví dụ
array_parent = [1, 2, 3, 4, 5, 6, 7, 8]
array_sub = [3, 4, 5]

result = is_subarray(array_sub, array_parent)
print(result)

```
